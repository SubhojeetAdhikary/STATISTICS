## Question 1

**Define Covariance and explain how it differs from Correlation in terms of scale and interpretation.**

### Solution

**Covariance** measures how two variables change together.

- A **positive covariance** means both variables generally move in the same direction.
- A **negative covariance** means they generally move in opposite directions.
- A covariance near **zero** suggests little linear co-movement.

The sample covariance formula is:

$$
Cov(X,Y)=\frac{\sum (x_i-\bar{x})(y_i-\bar{y})}{n-1}
$$

**Difference from correlation:** Covariance depends on the units and scale of the variables, so its magnitude is difficult to interpret directly. Correlation standardizes covariance and always lies between **-1 and +1**, making the strength and direction of the relationship easier to interpret.

## Question 2

**What does a positive, negative, and zero covariance indicate about the relationship between two variables?**

### Solution

- **Positive covariance:** When one variable increases, the other variable tends to increase.
- **Negative covariance:** When one variable increases, the other variable tends to decrease.
- **Zero covariance:** There is no clear linear co-movement between the two variables.

Zero covariance does not necessarily mean the variables are completely unrelated because a non-linear relationship may still exist.

## Question 3

**Discuss the limitations of covariance as a measure of relationship between two variables. Why is correlation preferred in many cases?**

### Solution

Covariance has some limitations:

1. Its value depends on the **units and scale** of the variables.
2. It has no fixed range, so its magnitude is difficult to interpret.
3. It mainly describes the direction of a **linear relationship**.
4. Covariance values from different datasets are difficult to compare directly.

Correlation is often preferred because it is standardized between **-1 and +1**, has no units, and clearly indicates both the **direction and strength** of a linear relationship.

## Question 4

**Explain the difference between Pearson's correlation coefficient and Spearman's rank correlation coefficient. When would you prefer to use Spearman's correlation?**

### Solution

**Pearson's correlation** measures the strength and direction of a **linear relationship** between two numerical variables.

**Spearman's rank correlation** measures the strength and direction of a **monotonic relationship** using the ranks of the observations instead of their original values.

Spearman's correlation is preferred when:

- The data is ordinal or ranked.
- The relationship is monotonic but not necessarily linear.
- The data contains strong outliers.
- Pearson's assumptions are not suitable for the data.

## Question 5

**If the correlation coefficient between two variables X and Y is 0.85, interpret this value in context. Can you infer causation from this value? Why or why not?**

### Solution

A correlation coefficient of **r = 0.85** indicates a **strong positive linear relationship** between X and Y. Generally, as X increases, Y also tends to increase.

However, we **cannot infer causation** from correlation alone. A high correlation only shows association. The relationship may be influenced by other variables, coincidence, or a common underlying factor. Therefore, **correlation does not imply causation**.

## Question 6

**Using the dataset below, calculate the covariance between X and Y.**

| X | 2 | 4 | 6 | 8 |
|---|---:|---:|---:|---:|
| Y | 3 | 7 | 5 | 10 |

### Solution

First calculate the means:

$$
\bar{X}=5
$$

$$
\bar{Y}=6.25
$$

Using the sample covariance formula:

$$
Cov(X,Y)=\frac{\sum (x_i-\bar{x})(y_i-\bar{y})}{n-1}
$$

The sum of the cross-products of deviations is **19**.

Therefore:

$$
Cov(X,Y)=\frac{19}{3}=6.33
$$

**Answer: Sample Covariance = 6.33**

The positive covariance indicates that X and Y generally move in the same direction.

```python
import numpy as np

X = np.array([2, 4, 6, 8])
Y = np.array([3, 7, 5, 10])

covariance = np.cov(X, Y, ddof=1)[0, 1]
print("Sample Covariance between X and Y:", round(covariance, 2))
```

## Question 7

**Compute the Pearson correlation coefficient between variables A and B:**

| A | 10 | 20 | 30 | 40 | 50 |
|---|---:|---:|---:|---:|---:|
| B | 8 | 14 | 18 | 24 | 28 |

### Solution

The Pearson correlation coefficient is calculated as:

$$
r=\frac{Cov(A,B)}{s_A s_B}
$$

For the given data:

**r ≈ 0.9976**

**Answer: Pearson correlation coefficient ≈ 0.998**

This indicates a **very strong positive linear correlation** between A and B.

```python
import numpy as np

A = np.array([10, 20, 30, 40, 50])
B = np.array([8, 14, 18, 24, 28])

correlation = np.corrcoef(A, B)[0, 1]
print("Pearson Correlation:", round(correlation, 4))
```

## Question 8

**The following table shows heights (in cm) and weights (in kg) of 5 students. Find the correlation coefficient between Height and Weight.**

| Height | 150 | 160 | 165 | 170 | 180 |
|---|---:|---:|---:|---:|---:|
| Weight | 50 | 55 | 58 | 62 | 70 |

### Solution

Using Pearson's correlation coefficient:

**r ≈ 0.9922**

**Answer: Correlation coefficient ≈ 0.992**

This shows a **very strong positive correlation** between height and weight in this dataset. As height increases, weight also tends to increase.

```python
import numpy as np

height = np.array([150, 160, 165, 170, 180])
weight = np.array([50, 55, 58, 62, 70])

correlation = np.corrcoef(height, weight)[0, 1]
print("Correlation between Height and Weight:", round(correlation, 4))
```

## Question 9

**Given the dataset below, determine whether there is a positive or negative correlation between X and Y. (No need for exact calculation, just reasoning.)**

| X | 1 | 2 | 3 | 4 | 5 |
|---|---:|---:|---:|---:|---:|
| Y | 15 | 12 | 9 | 7 | 3 |

### Solution

As X increases from **1 to 5**, Y decreases from **15 to 3**.

Therefore, X and Y have a **strong negative correlation**.

**Answer: Negative correlation.**

```python
import numpy as np

X = np.array([1, 2, 3, 4, 5])
Y = np.array([15, 12, 9, 7, 3])

correlation = np.corrcoef(X, Y)[0, 1]
print("Correlation:", round(correlation, 4))
print("The relationship is negative because Y decreases as X increases.")
```

## Question 10

**Two investment portfolios have the following returns (%) over 5 years. Compute the covariance and correlation coefficient, and interpret whether the portfolios move together.**

| Year | Portfolio A | Portfolio B |
|---|---:|---:|
| 1 | 8 | 6 |
| 2 | 10 | 9 |
| 3 | 12 | 11 |
| 4 | 9 | 8 |
| 5 | 11 | 10 |

### Solution

The average returns are:

$$
\bar{A}=10
$$

$$
\bar{B}=8.8
$$

Using sample covariance:

$$
Cov(A,B)=3.00
$$

The Pearson correlation coefficient is:

$$
r \approx 0.9864
$$

**Answer:**

- **Sample Covariance = 3.00**
- **Correlation coefficient ≈ 0.986**

The positive covariance and correlation close to +1 indicate that the two portfolios have a **very strong positive relationship** and generally move together in the same direction.

```python
import numpy as np

portfolio_A = np.array([8, 10, 12, 9, 11])
portfolio_B = np.array([6, 9, 11, 8, 10])

covariance = np.cov(portfolio_A, portfolio_B, ddof=1)[0, 1]
correlation = np.corrcoef(portfolio_A, portfolio_B)[0, 1]

print("Sample Covariance:", round(covariance, 2))
print("Pearson Correlation:", round(correlation, 4))
```
