## **Q3: Mode in Real Life**
A store records the shoe sizes sold in one day: [7, 8, 9, 8, 8, 10, 7, 9].

What is the mode, and why is this information useful for the store manager?

## **Ans-**

### **Step 1: Find the Mode**

Data: [7, 8, 9, 8, 8, 10, 7, 9]

Count the frequency of each shoe size:

Size 7 → 2 times

Size 8 → 3 times

Size 9 → 2 times

Size 10 → 1 time

The shoe size that appears most often is 8.

**Mode = 8**

### **Step 2: Why is this useful?**

The mode tells the store manager which shoe size is sold most frequently. Since size 8 is the most popular size, the manager can:

- Keep more size 8 shoes in stock.
- Reorder size 8 more often.
- Reduce the risk of running out of the most demanded size.
- Plan inventory more efficiently.

 Which company is more consistent, and why?

## **Ans-**

Company A: Variance = 6

## **Q9: Combining Measures**
A family records their monthly electricity usage (in kWh): [400, 420, 390, 450, 410].

Find the mean and standard deviation. What do these values together tell you about the family’s energy use
pattern?

## **Ans-**

Monthly Electricity Usage (kWh): [400, 420, 390, 450, 410]

### **Step 1: Calculate the Mean**

Mean = (400 + 420 + 390 + 450 + 410) / 5

     = 2070 / 5
     = 414 kWh

Step 2: Calculate the Squared Deviations from the Mean

400 - 414 = -14   → (-14)² = 196

420 - 414 = 6     → (6)² = 36

390 - 414 = -24   → (-24)² = 576

450 - 414 = 36    → (36)² = 1296

410 - 414 = -4    → (-4)² = 16

Sum of Squared Deviations

= 196 + 36 + 576 + 1296 + 16
= 2120

### **Step 3: Calculate the Variance**

Variance = 2120 / 5

         = 424

### **Step 4: Calculate the Standard Deviation**

Standard Deviation = √424

                   ≈ 20.6 kWh

**Results:**

Mean = 414 kWh

Standard Deviation ≈ 20.6 kWh

**Interpretation:**

Interpretation:
The family uses an average of 414 kWh of electricity per month. The standard deviation of 20.6 kWh indicates that monthly usage varies only moderately, suggesting a fairly consistent energy consumption pattern.

## **Q10: Practical Application**
A basketball player’s points in 8 games are recorded: [15, 18, 20, 22, 25, 17, 19, 21].

Find the mean, median, mode, range, and standard deviation. What insights can these measures provide
about the player’s scoring performance?

## **Ans-**

Basketball Player's Points:
[15, 18, 20, 22, 25, 17, 19, 21]

### **Step 1: Calculate the Mean**

Mean = (15 + 18 + 20 + 22 + 25 + 17 + 19 + 21) / 8

     = 157 / 8
     = 19.625

### **Step 2: Calculate the Median**

Arrange the data in ascending order:
[15, 17, 18, 19, 20, 21, 22, 25]

Since there are 8 values (even number),

Median = (4th value + 5th value) / 2

       = (19 + 20) / 2
       = 19.5

### **Step 3: Calculate the Mode**

Each value appears only once.

Mode = No mode

### **Step 4: Calculate the Range**

Range = Maximum Value - Minimum Value

      = 25 - 15
      = 10

### **Step 5: Calculate the Standard Deviation**

Mean = 19.625

Squared Deviations:

(15 - 19.625)² = 21.390625

(18 - 19.625)² = 2.640625

(20 - 19.625)² = 0.140625

(22 - 19.625)² = 5.640625

(25 - 19.625)² = 28.890625

(17 - 19.625)² = 6.890625

(19 - 19.625)² = 0.390625

(21 - 19.625)² = 1.890625

Sum of Squared Deviations = 67.875

Variance = 67.875 / 8

         = 8.484375

Standard Deviation = √8.484375

                   ≈ 2.91

**Results:**

Mean = 19.63

Median = 19.5

Mode = No mode

Range = 10

Standard Deviation ≈ 2.91

**Interpretation:**

The player scores about 19.63 points per game on average. The median of 19.5 is close to the mean, indicating a fairly balanced distribution of scores. There is no mode because no score repeats. The range of 10 points shows the difference between the highest and lowest scoring games. The standard deviation of 2.91 suggests that the player's scoring is fairly consistent, with only moderate variation from game to game.
