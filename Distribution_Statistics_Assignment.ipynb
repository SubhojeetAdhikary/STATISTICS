## Question 1

**Simulate 30 rolls with `=RANDBETWEEN(1,6)`. What is the probability of rolling a 3 exactly 5 times? (Hint: Use BINOM.DIST)**

### Solution

For a fair die:

- Number of trials, **n = 30**
- Probability of rolling a 3 in one roll, **p = 1/6**
- Required number of successes, **x = 5**

Using the binomial probability formula:

\[
P(X=5)=\binom{30}{5}\left(\frac{1}{6}\right)^5\left(\frac{5}{6}\right)^{25}
\]

In Excel, the formula is:

`=BINOM.DIST(5,30,1/6,FALSE)`

Therefore,

**Probability = 0.1921 ≈ 19.21%**.

```python
import numpy as np
from math import comb

# Simulate 30 die rolls
rolls = np.random.randint(1, 7, 30)
count_of_3 = np.sum(rolls == 3)

# Exact binomial probability
probability = comb(30, 5) * (1/6)**5 * (5/6)**25

print("Simulated rolls:", rolls)
print("Number of 3s in this simulation:", count_of_3)
print("Probability of exactly 5 threes:", round(probability, 4))
print("Percentage:", round(probability * 100, 2), "%")
```

## Question 2

**Generate 100 values in Excel using the continuous uniform distribution `RAND()` and plot a histogram. Describe the shape of the distribution.**

### Solution

`RAND()` generates random values between **0 and 1**, where every value in this interval has an equal chance of occurring.

The histogram of 100 random values should appear **approximately uniform or rectangular**, although small variations may occur because only 100 observations are used.

In Excel, 100 values can be generated by entering `=RAND()` and filling it down for 100 rows, then inserting a Histogram chart.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 100 values from a continuous uniform distribution
values = np.random.rand(100)

plt.figure(figsize=(8, 5))
plt.hist(values, bins=10, edgecolor='black')
plt.title("Histogram of 100 Uniform Random Values")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Question 3

**A dataset has a mean of 50 and a standard deviation of 5. What percentage of values lie between 45 and 55 if the data follows a normal distribution?**

### Solution

Given:

- Mean, **μ = 50**
- Standard deviation, **σ = 5**
- Lower value = 45
- Upper value = 55

The interval 45 to 55 is exactly **one standard deviation below and above the mean**:

\[
50 - 5 = 45
\]

\[
50 + 5 = 55
\]

According to the Empirical Rule for a normal distribution, approximately **68.27%** of observations lie within one standard deviation of the mean.

**Answer: Approximately 68.27% of the values lie between 45 and 55.**

## Question 4

**What is the concept of standardization (z-score), and why is it important in data analysis? Explain the formula and how standardization transforms a dataset.**

### Solution

Standardization converts a value into a **z-score**, which tells us how many standard deviations that value is above or below the mean.

The formula is:

\[
z = \frac{x-\mu}{\sigma}
\]

Where:

- **x** = original value
- **μ** = mean
- **σ** = standard deviation
- **z** = standardized value

After standardization, the transformed dataset has:

- **Mean = 0**
- **Standard deviation = 1**

Standardization is important because it puts variables measured on different scales onto a common scale. This makes comparison easier and is useful in many statistical and machine-learning techniques.

## Question 5

**What is Kurtosis and their type?**

### Solution

Kurtosis measures the **shape of the tails and peak of a distribution** compared with a normal distribution. It helps identify whether the data has more or fewer extreme values.

The main types of kurtosis are:

1. **Mesokurtic** – Similar to a normal distribution. Excess kurtosis is approximately 0.
2. **Leptokurtic** – Has heavier tails and more extreme values than a normal distribution. Excess kurtosis is positive.
3. **Platykurtic** – Has lighter tails and fewer extreme values than a normal distribution. Excess kurtosis is negative.

## Question 6

**Explain why the uniform distribution is a good model for the outcome of rolling a fair die.**

### Solution

A fair die has six possible outcomes: **1, 2, 3, 4, 5, and 6**.

Each outcome has the same probability:

\[
P(X=x)=\frac{1}{6}
\]

Because every possible outcome is equally likely, the result of a fair die follows a **discrete uniform distribution**. Therefore, the uniform distribution is an appropriate model for a fair die.

## Question 7

**Use Excel to compute the probability of getting at least 8 successes in 15 trials with success probability 0.5.**

### Solution

Given:

- Number of trials, **n = 15**
- Probability of success, **p = 0.5**
- Required result: **at least 8 successes**

We need:

\[
P(X \geq 8)
\]

In Excel, use:

`=1-BINOM.DIST(7,15,0.5,TRUE)`

This calculates 1 minus the probability of getting 0 to 7 successes.

Therefore,

**P(X ≥ 8) = 0.5 = 50%**.

```python
from math import comb

prob_at_least_8 = sum(comb(15, k) * (0.5**k) * (0.5**(15-k)) for k in range(8, 16))

print("Probability of at least 8 successes:", prob_at_least_8)
print("Percentage:", prob_at_least_8 * 100, "%")
```

## Question 8

**How does log transformation help in stabilizing variance and making data more normally distributed?**

### Solution

A log transformation replaces each positive value **x** with **log(x)**.

It is useful for datasets that are strongly **right-skewed** or where larger values have much greater variability than smaller values.

Log transformation helps by:

- Compressing very large values more than small values.
- Reducing the effect of extreme observations.
- Reducing right skewness.
- Making the variance more stable across the dataset.
- Often making the distribution closer to a normal distribution.

For example, values such as `10, 100, 1000` become approximately `1, 2, 3` when using log base 10. This greatly reduces the difference between very large observations.
