## Question 1

**What is a null hypothesis (H₀) and why is it important in hypothesis testing?**

### Solution

A **null hypothesis (H₀)** is a statement that assumes there is **no significant difference, effect, or relationship** in the population.

For example:

**H₀: The average marks of two classes are equal.**

It is important because hypothesis testing starts by assuming that H₀ is true. The sample data is then used to decide whether there is enough statistical evidence to reject it.

## Question 2

**What does the significance level (α) represent in hypothesis testing?**

### Solution

The **significance level (α)** is the maximum probability of making a **Type I error**, which means rejecting a true null hypothesis.

A commonly used significance level is:

**α = 0.05**

This means we are willing to accept a **5% risk** of rejecting H₀ when it is actually true.

Decision rule:

- If **p-value < α**, reject H₀.
- If **p-value ≥ α**, fail to reject H₀.

## Question 3

**Differentiate between Type I and Type II errors.**

### Solution

| Error | Meaning |
|---|---|
| **Type I Error** | Rejecting H₀ when H₀ is actually true. |
| **Type II Error** | Failing to reject H₀ when H₀ is actually false. |

The probability of a Type I error is represented by **α**, while the probability of a Type II error is represented by **β**.

## Question 4

**Explain the difference between a one-tailed and two-tailed test. Give an example of each.**

### Solution

A **one-tailed test** checks for a difference in only one specific direction.

Example:

- H₀: μ = 50
- H₁: μ > 50

Here, we only want to test whether the mean is **greater than 50**.

A **two-tailed test** checks for a difference in either direction.

Example:

- H₀: μ = 50
- H₁: μ ≠ 50

Here, we want to test whether the mean is either **greater than or less than 50**.

## Question 5

**A company claims that the average time to resolve a customer complaint is 10 minutes.  
A random sample of 9 complaints gives an average time of 12 minutes and a standard deviation of 3 minutes.  
At α = 0.05, test the claim.**

### Solution

Since the population standard deviation is not given and the sample size is small, a **one-sample t-test** is appropriate.

### Step 1: State the hypotheses

- **H₀: μ = 10 minutes**
- **H₁: μ ≠ 10 minutes**

This is a **two-tailed test**.

### Step 2: Calculate the test statistic

Given:

- Sample mean, x̄ = 12
- Population mean under H₀, μ = 10
- Sample standard deviation, s = 3
- Sample size, n = 9

Formula:

\[
t = \frac{\bar{x}-\mu}{s/\sqrt{n}}
\]

\[
t = \frac{12-10}{3/\sqrt{9}} = 2.00
\]

Degrees of freedom:

\[
df = n-1 = 8
\]

At α = 0.05 for a two-tailed test, the critical t-value is approximately **±2.306**.

The p-value is approximately **0.0805**.

### Decision

Since **p-value = 0.0805 > 0.05**, we **fail to reject H₀**.

### Conclusion

There is **not enough statistical evidence** at the 5% significance level to conclude that the average complaint resolution time is different from 10 minutes. Therefore, the company's claim cannot be rejected.

```python
from scipy import stats
from math import sqrt

n = 9
sample_mean = 12
claimed_mean = 10
sample_sd = 3

t_stat = (sample_mean - claimed_mean) / (sample_sd / sqrt(n))
df = n - 1
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df))

print("t-statistic:", round(t_stat, 3))
print("Degrees of freedom:", df)
print("p-value:", round(p_value, 4))
```

## Question 6

**When should you use a z-test instead of a t-test?**

### Solution

A **z-test** is generally used when:

- The **population standard deviation (σ) is known**, and
- The sample size is sufficiently large, commonly **n ≥ 30**.

A **t-test** is usually used when:

- The population standard deviation is unknown, especially for small samples.
- The sample standard deviation is used as an estimate of the population standard deviation.

## Question 7

**The productivity of 6 employees was measured before and after a training program.**

| Employee | Before | After |
|---|---:|---:|
| 1 | 50 | 55 |
| 2 | 60 | 65 |
| 3 | 58 | 59 |
| 4 | 55 | 58 |
| 5 | 62 | 63 |
| 6 | 56 | 59 |

**At α = 0.05, test if the training improved productivity.**

### Solution

Because the measurements are taken from the **same employees before and after training**, a **paired t-test** is appropriate.

### Step 1: State the hypotheses

- **H₀: The training did not significantly change productivity.**
- **H₁: The training significantly improved productivity.**

The differences (After − Before) are:

**5, 5, 1, 3, 1, 3**

Mean difference:

\[
\bar{d} = 3.00
\]

Standard deviation of differences:

\[
s_d = 1.79
\]

Test statistic:

\[
t = \frac{\bar{d}}{s_d/\sqrt{n}} = 4.108
\]

Degrees of freedom:

\[
df = 6-1 = 5
\]

The two-tailed p-value is approximately **0.0093**.

### Decision

Since **p-value < 0.05**, we **reject H₀**.

### Conclusion

There is statistically significant evidence that the training program **improved employee productivity**.

```python
from scipy import stats

before = [50, 60, 58, 55, 62, 56]
after  = [55, 65, 59, 58, 63, 59]

t_stat, p_two_tailed = stats.ttest_rel(after, before)

print("Paired t-statistic:", round(t_stat, 3))
print("Two-tailed p-value:", round(p_two_tailed, 4))
print("One-tailed p-value for improvement:", round(p_two_tailed / 2, 4))
```

## Question 8

**A company wants to test if product preference is independent of gender.**

| Gender | Product A | Product B | Total |
|---|---:|---:|---:|
| Male | 30 | 20 | 50 |
| Female | 10 | 40 | 50 |
| **Total** | **40** | **60** | **100** |

**At α = 0.05, test independence.**

### Solution

A **Chi-square test of independence** is used because both variables are categorical.

### Step 1: State the hypotheses

- **H₀: Product preference is independent of gender.**
- **H₁: Product preference is associated with gender.**

Expected frequencies are calculated using:

\[
E = \frac{(Row\ Total)(Column\ Total)}{Grand\ Total}
\]

Expected table:

| Gender | Product A | Product B |
|---|---:|---:|
| Male | 20 | 30 |
| Female | 20 | 30 |

The Chi-square statistic is:

\[
\chi^2 = 16.667
\]

Degrees of freedom:

\[
df = (2-1)(2-1) = 1
\]

The p-value is approximately **0.000045**.

### Decision

Since **p-value < 0.05**, we **reject H₀**.

### Conclusion

There is sufficient statistical evidence to conclude that **product preference is not independent of gender**. In other words, gender and product preference are significantly associated.

```python
from scipy.stats import chi2_contingency
import numpy as np

observed = np.array([
    [30, 20],
    [10, 40]
])

chi2, p_value, dof, expected = chi2_contingency(observed, correction=False)

print("Chi-square statistic:", round(chi2, 3))
print("Degrees of freedom:", dof)
print("p-value:", round(p_value, 6))
print("\nExpected frequencies:")
print(expected)
```
